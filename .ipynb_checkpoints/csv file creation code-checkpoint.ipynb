```python
import os
import pandas as pd

dataset_path = "C:/Users/DELL 3490/IdeaProjects/HARP/Dataset"  # path to your dataset folder

data = []

# Loop through class folders
for label in os.listdir(dataset_path):
    class_folder = os.path.join(dataset_path, label)
    
    if os.path.isdir(class_folder):  # Check if folder
        for image_name in os.listdir(class_folder):
            
            if image_name.endswith(('.png', '.jpg', '.jpeg')):
                
                # Remove extension
                image_id = os.path.splitext(image_name)[0]
                
                data.append([image_id, int(label)])

# Create DataFrame
df = pd.DataFrame(data, columns=["id_code", "diagnosis"])

# Save CSV file
df.to_csv("train_1.csv", index=False)

print("CSV file created successfully!")
print(df.head())
```
